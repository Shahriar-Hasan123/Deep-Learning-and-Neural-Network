```python


def parse_wider_annotations(txt_path, img_dir):
    """Parse WIDER FACE annotation file"""
    print(f"Parsing annotations from: {txt_path}")
    print(f"Image directory: {img_dir}")

    with open(txt_path, 'r') as f:
        lines = [line.strip() for line in f if line.strip()]

    data = []
    i = 0
    skipped = 0

    while i < len(lines):
        img_path = lines[i]
        i += 1
        if i >= len(lines):
            break

        try:
            num_faces = int(lines[i])
        except ValueError:
            print(f"Skipping malformed entry at line {i}: {lines[i]}")
            skipped += 1
            continue
        i += 1

        boxes = []
        for _ in range(num_faces):
            if i >= len(lines):
                break
            parts = lines[i].split()
            if len(parts) >= 4:
                x, y, w, h = map(int, parts[:4])
                # Skip invalid boxes
                if w > 0 and h > 0:
                    boxes.append([x, y, w, h])
            i += 1

        # Only add images with valid boxes
        if boxes:
            full_img_path = os.path.join(img_dir, img_path)
            if os.path.exists(full_img_path):
                data.append((full_img_path, boxes))
            else:
                skipped += 1

    print(f"Loaded {len(data)} images, skipped {skipped}")
    return data



def build_yolov1(input_shape=(448, 448, 3), S=7, B=2, C=1):
    """Build YOLOv1-inspired architecture for face detection (memory-efficient)"""
    inputs = layers.Input(shape=input_shape)

    # Convolutional layers - reduced filters
    x = layers.Conv2D(32, 7, strides=2, padding='same')(inputs)
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.MaxPooling2D(2, strides=2)(x)

    x = layers.Conv2D(64, 3, padding='same')(x)
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.MaxPooling2D(2, strides=2)(x)

    x = layers.Conv2D(128, 3, padding='same')(x)
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.Conv2D(64, 1, padding='same')(x)
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.Conv2D(128, 3, padding='same')(x)
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.MaxPooling2D(2, strides=2)(x)

    x = layers.Conv2D(256, 3, padding='same')(x)
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.Conv2D(128, 1, padding='same')(x)
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.Conv2D(256, 3, padding='same')(x)
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.MaxPooling2D(2, strides=2)(x)

    x = layers.Conv2D(512, 3, padding='same')(x)
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.Conv2D(256, 1, padding='same')(x)
    x = layers.LeakyReLU(alpha=0.1)(x)

    # Fully connected layers - drastically reduced
    x = layers.Flatten()(x)
    x = layers.Dense(512)(x)  # Reduced from 4096 to 512
    x = layers.LeakyReLU(alpha=0.1)(x)
    x = layers.Dropout(0.5)(x)
    x = layers.Dense(S * S * (B * 5 + C))(x)

    outputs = layers.Reshape((S, S, B * 5 + C))(x)

    model = models.Model(inputs, outputs)
    return model


def preprocess_image(img_path, boxes, target_size=(448, 448), S=7, B=2, C=1):
    """Preprocess image and create YOLO label tensor"""
    img = cv2.imread(img_path)
    if img is None:
        return None, None

    orig_h, orig_w = img.shape[:2]
    img = cv2.resize(img, target_size)
    img = img / 255.0

    label = np.zeros((S, S, B * 5 + C))

    for box in boxes:
        x, y, bw, bh = box


        cx = (x + bw / 2) / orig_w
        cy = (y + bh / 2) / orig_h
        bw_norm = bw / orig_w
        bh_norm = bh / orig_h


        grid_x = int(cx * S)
        grid_y = int(cy * S)


        grid_x = max(0, min(S - 1, grid_x))
        grid_y = max(0, min(S - 1, grid_y))


        x_cell = (cx * S) - grid_x
        y_cell = (cy * S) - grid_y


        if label[grid_y, grid_x, 4] == 0:

            label[grid_y, grid_x, 0] = x_cell
            label[grid_y, grid_x, 1] = y_cell
            label[grid_y, grid_x, 2] = bw_norm
            label[grid_y, grid_x, 3] = bh_norm
            label[grid_y, grid_x, 4] = 1.0  # confidence

            # Set class (face = 1)
            if C > 0:
                label[grid_y, grid_x, B * 5] = 1.0

    return img, label


def data_generator(data, batch_size=8, S=7, B=2, C=1):
    """Generate batches of training data"""
    indices = np.arange(len(data))

    while True:
        np.random.shuffle(indices)

        for start_idx in range(0, len(data), batch_size):
            batch_indices = indices[start_idx:start_idx + batch_size]
            batch_imgs, batch_labels = [], []

            for idx in batch_indices:
                img_path, boxes = data[idx]
                img, label = preprocess_image(img_path, boxes, S=S, B=B, C=C)

                if img is not None:
                    batch_imgs.append(img)
                    batch_labels.append(label)

            if len(batch_imgs) > 0:
                yield np.array(batch_imgs), np.array(batch_labels)



def yolo_loss(y_true, y_pred, lambda_coord=5.0, lambda_noobj=0.5):
    """YOLOv1 loss function"""
    obj_mask = y_true[..., 4:5]
    noobj_mask = 1 - obj_mask

    coord_loss = lambda_coord * tf.reduce_sum(
        obj_mask * tf.square(y_true[..., 0:2] - y_pred[..., 0:2])
    )

    size_loss = lambda_coord * tf.reduce_sum(
        obj_mask * tf.square(
            tf.sqrt(tf.abs(y_true[..., 2:3]) + 1e-6) -
            tf.sqrt(tf.abs(y_pred[..., 2:3]) + 1e-6)
        )
    )
    size_loss += lambda_coord * tf.reduce_sum(
        obj_mask * tf.square(
            tf.sqrt(tf.abs(y_true[..., 3:4]) + 1e-6) -
            tf.sqrt(tf.abs(y_pred[..., 3:4]) + 1e-6)
        )
    )

    obj_conf_loss = tf.reduce_sum(
        obj_mask * tf.square(y_true[..., 4:5] - y_pred[..., 4:5])
    )
    noobj_conf_loss = lambda_noobj * tf.reduce_sum(
        noobj_mask * tf.square(y_true[..., 4:5] - y_pred[..., 4:5])
    )

    # Class loss
    class_loss = tf.reduce_sum(
        obj_mask * tf.square(y_true[..., 5:] - y_pred[..., 5:])
    )

    total_loss = coord_loss + size_loss + obj_conf_loss + noobj_conf_loss + class_loss
    return total_loss


train_data = parse_wider_annotations(
    os.path.join(path, "wider_face_annotations/wider_face_split/wider_face_train_bbx_gt.txt"),
    os.path.join(path, "WIDER_train/WIDER_train/images")
)

val_data = parse_wider_annotations(
    os.path.join(path, "wider_face_annotations/wider_face_split/wider_face_val_bbx_gt.txt"),
    os.path.join(path, "WIDER_val/WIDER_val/images")
)

print(f"\nTraining samples: {len(train_data)}")
print(f"Validation samples: {len(val_data)}")


gpus = tf.config.experimental.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
        print(f"GPU memory growth enabled for {len(gpus)} GPU(s)")
    except RuntimeError as e:
        print(e)

print("\nBuilding model...")
model = build_yolov1(S=7, B=2, C=1)
model.summary()

print("\nCompiling model...")
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss=yolo_loss
)


train_gen = data_generator(train_data, batch_size=4)  # Reduced from 16 to 4
val_gen = data_generator(val_data, batch_size=4)

# Training callbacks
callbacks = [
    tf.keras.callbacks.ModelCheckpoint(
        'yolov1_face_best.h5',
        save_best_only=True,
        monitor='val_loss',
        verbose=1
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        min_lr=1e-7,
        verbose=1
    ),
    tf.keras.callbacks.EarlyStopping(
        monitor='val_loss',
        patience=5,
        verbose=1,
        restore_best_weights=True
    )
]

print("\nStarting training...")
history = model.fit(
    train_gen,
    steps_per_epoch=200,
    validation_data=val_gen,
    validation_steps=40,
    epochs=30,
    callbacks=callbacks
)

# Plot training history
plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training History')
plt.legend()
plt.grid(True)
plt.show()


def draw_boxes(img, preds, S=7, threshold=0.3):
    """Draw predicted bounding boxes on image"""
    h, w, _ = img.shape

    for i in range(S):
        for j in range(S):
            conf = preds[i, j, 4]

            if conf > threshold:
                x_cell = preds[i, j, 0]
                y_cell = preds[i, j, 1]
                bw = preds[i, j, 2]
                bh = preds[i, j, 3]

                cx = (j + x_cell) / S
                cy = (i + y_cell) / S

                x1 = int((cx - bw / 2) * w)
                y1 = int((cy - bh / 2) * h)
                x2 = int((cx + bw / 2) * w)
                y2 = int((cy + bh / 2) * h)

                x1 = max(0, min(w-1, x1))
                y1 = max(0, min(h-1, y1))
                x2 = max(0, min(w-1, x2))
                y2 = max(0, min(h-1, y2))

                cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
                cv2.putText(img, f"{conf:.2f}", (x1, y1 - 5),
                           cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    return img


print("\nTesting on sample images...")

sample_indices = np.random.choice(len(val_data), min(5, len(val_data)), replace=False)

plt.figure(figsize=(15, 10))
for idx, sample_idx in enumerate(sample_indices):
    img_path, boxes = val_data[sample_idx]

    # Load and preprocess
    img = cv2.imread(img_path)
    img_resized = cv2.resize(img, (448, 448)) / 255.0

    # Predict
    preds = model.predict(np.expand_dims(img_resized, axis=0), verbose=0)[0]

    # Draw boxes
    img_with_boxes = draw_boxes(img.copy(), preds, threshold=0.3)

    # Display
    plt.subplot(2, 3, idx + 1)
    plt.imshow(cv2.cvtColor(img_with_boxes, cv2.COLOR_BGR2RGB))
    plt.title(f'Sample {idx + 1}')
    plt.axis('off')

plt.tight_layout()
plt.show()


print("\nTest on the uploaded image...")
import urllib.request

image_url = "https://static-bestcolleges.tosshub.com/2025/News/SNfKTVxUahOJF03a1MAy8cRKsDGQMeI2nCSe8DgV.jpg"
image_path = "test_image.jpg"

try:
    urllib.request.urlretrieve(image_url, image_path)

    # Load and predict
    img = cv2.imread(image_path)
    img_resized = cv2.resize(img, (448, 448)) / 255.0
    preds = model.predict(np.expand_dims(img_resized, axis=0), verbose=0)[0]

    # Draw boxes
    img_with_boxes = draw_boxes(img.copy(), preds, threshold=0.3)

    # Display
    plt.figure(figsize=(12, 8))
    plt.imshow(cv2.cvtColor(img_with_boxes, cv2.COLOR_BGR2RGB))
    plt.title('Face Detection on Test Image')
    plt.axis('off')
    plt.show()

    print("Detection complete!")
except Exception as e:
    print(f"Error loading test image: {e}")


# Save final model
model.save('yolov1_face_final.h5')
print("\nModel saved as 'yolov1_face_final.h5'")
```
