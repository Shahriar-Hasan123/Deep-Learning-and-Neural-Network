```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.applications import VGG16
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.models import Model
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA

# Step 1: Load and Resize MNIST
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Resize to (48, 48, 3)
x_train = np.stack([x_train]*3, axis=-1)
x_test = np.stack([x_test]*3, axis=-1)

x_train = tf.image.resize(x_train, [48, 48]).numpy()
x_test = tf.image.resize(x_test, [48, 48]).numpy()

x_train = x_train / 255.0
x_test = x_test / 255.0


# Step 2: Data Generator

train_datagen = ImageDataGenerator()
test_datagen = ImageDataGenerator()

batch_size = 128

train_generator = train_datagen.flow(x_train, y_train, batch_size=batch_size, shuffle=True)
test_generator = test_datagen.flow(x_test, y_test, batch_size=batch_size, shuffle=False)


# Step 3: Pretrained VGG16

base_model = VGG16(weights="imagenet", include_top=False, input_shape=(48,48,3))
x = GlobalAveragePooling2D()(base_model.output)
feature_extractor = Model(inputs=base_model.input, outputs=x)

# Features BEFORE transfer learning
features_before = feature_extractor.predict(x_test, batch_size=batch_size)


# Step 4: Transfer Learning on MNIST

for layer in base_model.layers:
    layer.trainable = False

x = GlobalAveragePooling2D()(base_model.output)
x = Dense(256, activation="relu")(x)
x = Dropout(0.5)(x)
output = Dense(10, activation="softmax")(x)

model = Model(inputs=base_model.input, outputs=output)
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])

model.fit(train_generator, epochs=5, validation_data=test_generator)

# Features AFTER transfer learning
extractor_after = Model(inputs=model.input, outputs=model.layers[-3].output)
features_after = extractor_after.predict(x_test, batch_size=batch_size)


# Step 5: Visualization Function

def plot_side_by_side(method_name, features_before, features_after, y_test, reducer_before, reducer_after, sample_size=None):
    if sample_size:  # For t-SNE to reduce time
        idx = np.random.choice(len(features_before), sample_size, replace=False)
        features_before, y_b = features_before[idx], y_test[idx]
        features_after, y_a = features_after[idx], y_test[idx]
    else:
        y_b, y_a = y_test, y_test

    reduced_before = reducer_before.fit_transform(features_before, y_b if method_name=="LDA" else None)
    reduced_after = reducer_after.fit_transform(features_after, y_a if method_name=="LDA" else None)

    fig, axes = plt.subplots(1, 2, figsize=(12,5))
    sc1 = axes[0].scatter(reduced_before[:,0], reduced_before[:,1], c=y_b, cmap="tab10", s=10)
    axes[0].set_title(f"{method_name} - Before Transfer")
    sc2 = axes[1].scatter(reduced_after[:,0], reduced_after[:,1], c=y_a, cmap="tab10", s=10)
    axes[1].set_title(f"{method_name} - After Transfer")

    plt.colorbar(sc1, ax=axes, orientation="horizontal", fraction=0.05)
    plt.suptitle(f"{method_name} Comparison", fontsize=14)
    plt.show()


# Step 6: Plot Comparisons

# PCA
plot_side_by_side("PCA", features_before, features_after, y_test, PCA(n_components=2), PCA(n_components=2))

# t-SNE (use subset to avoid long runtime)
plot_side_by_side("t-SNE", features_before, features_after, y_test,
                  TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000),
                  TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000),
                  sample_size=2000)

# LDA
plot_side_by_side("LDA", features_before, features_after, y_test, LDA(n_components=2), LDA(n_components=2))
```
